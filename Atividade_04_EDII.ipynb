```python
class FilaPrioridadeOrdenada:
    def __init__(self):
        self.fila = []

    def add(self, item):
        self.fila.append(item)
        self.fila.sort(key=lambda x: x[0])

    def remove_min(self):
        if not self.fila:
            return None
        return self.fila.pop(0)

    def __str__(self):
        return str(self.fila)


class FilaPrioridadeNaoOrdenada:
    def __init__(self):
        self.fila = []

    def add(self, item):
        self.fila.append(item)

    def remove_min(self):
        if not self.fila:
            return None
        menor = 0
        for i in range(1, len(self.fila)):
            if self.fila[i][0] < self.fila[menor][0]:
                menor = i
        return self.fila.pop(menor)

    def __str__(self):
        return str(self.fila)


comandos = [
    ("add", (8, "G")),
    ("add", (7, "F")),
    ("add", (5, "A")),
    ("remove_min", None),
    ("remove_min", None),
    ("add", (4, "B")),
    ("add", (1, "D")),
    ("remove_min", None),
    ("add", (3, "J")),
    ("add", (6, "L")),
    ("remove_min", None),
    ("add", (2, "H")),
    ("remove_min", None),
    ("remove_min", None)
]

fila_ord = FilaPrioridadeOrdenada()
fila_nao_ord = FilaPrioridadeNaoOrdenada()

for acao, valor in comandos:
    if acao == "add":
        fila_ord.add(valor)
        fila_nao_ord.add(valor)
    else:
        fila_ord.remove_min()
        fila_nao_ord.remove_min()

print("Fila ordenada final:", fila_ord)
print("Fila não ordenada final:", fila_nao_ord)
```
